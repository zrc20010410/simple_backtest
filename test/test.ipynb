```python
from event import SignalEvent
# signals = pd.read_csv(path, dtype={"datetime": "string","symbol": "string", "quantity": "Int64"})
path = DEFAULT_DATA_DIR / 'signals.csv'
signals = pd.read_csv(path, dtype={"datetime": "string","symbol": "string", "quantity": "Int64"})
signals
```

```python
signals = pd.read_csv(path, dtype={"datetime": str, "symbol": str})
events_by_date: dict[str, list[SignalEvent]] = {}
for row in signals.itertuples(index=False):
    event = SignalEvent(
        datetime=row.datetime,
        symbol=row.symbol,
        signal_type=row.signal_type,
        quantity=row.quantity,
    )
    # events_by_date.setdefault(row.datetime, []).append(event)
    events_by_date.setdefault(row.datetime, []).append(event)
events_by_date
```
