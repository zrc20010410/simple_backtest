```python
from pathlib import Path
import pandas as pd

DEFAULT_SIGNAL_FILE = 'E:/Code/quant/src/backtester/data/signals.csv'
SIGNAL_COLUMNS = ["datetime", "symbol", "signal_type", "quantity"]

class Event(object):
    pass

class SignalEvent(Event):

    def __init__(self, datetime: str, symbol: str, signal_type: str, quantity: int):
        """
        datetime - Signal 产生的时间
        symbol - 交易标的
        signal_type - 'BUY' or 'SELL'
        quantity - 交易数量
        """

        self.type = 'SIGNAL'
        self.symbol = symbol
        self.datetime = datetime
        self.signal_type = signal_type
        self.quantity = quantity

    @staticmethod
    def save_csv(signals: pd.DataFrame, path=DEFAULT_SIGNAL_FILE) -> Path:
        """将整张信号表写入 CSV，覆盖原文件。"""
        path = Path(path)
        path.parent.mkdir(parents=True, exist_ok=True)
        signals.to_csv(path, columns=SIGNAL_COLUMNS, index=False)
        return path

    @classmethod
    def read_csv(cls, path=DEFAULT_SIGNAL_FILE):
        """按 CSV 的原始行序还原信号，日期和股票代码保留为字符串。"""
        signals = pd.read_csv(path, dtype=str, keep_default_na=False)
        return [
            cls(date, symbol, signal_type, int(quantity))
            for date, symbol, signal_type, quantity
            in signals[SIGNAL_COLUMNS].itertuples(index=False, name=None)
        ]
```
